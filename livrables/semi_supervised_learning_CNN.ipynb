# Étape 4 - Apprentissage semi-supervisé (variante CNN avec fine-tuning)

**Projet BrainScanAI** - Détection de tumeurs cérébrales  

**Architecture** : ResNet50 pré-entraîné avec **fine-tuning des 2 dernières couches** :
- `layer4` (dernier bloc résiduel convolutif) — dégelé, entraîné avec un LR faible
- `fc` remplacée par une tête MLP (2048 → 512 → 128 → 2) — entraînée avec un LR plus élevé
- `layer1`, `layer2`, `layer3`, `conv1` — gelés (poids ImageNet conservés)

**Plan du notebook** :
1. Chargement des images et transforms
2. **HP tuning** : recherche des meilleurs hyperparamètres (tête fc + lr de layer4)
3. Entraînement des modèles A / B / C avec les meilleurs HP
4. Comparaison avec la version MLP

## 4. Transforms

ResNet50 attend des images **224×224 normalisées avec les statistiques ImageNet**.  
Même preprocessing que lors de l'extraction des features à l'étape 2.

```python
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

# Transform standard : redimensionnement + crop centré + normalisation ImageNet
transform_standard = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
])

# Transform avec augmentation : variations aléatoires pour enrichir le dataset du modèle C
transform_aug = transforms.Compose([
    transforms.Resize(256),
    transforms.RandomCrop(224),              # crop aléatoire (vs centré)
    transforms.RandomHorizontalFlip(p=0.5),  # flip horizontal 50% du temps
    transforms.RandomRotation(degrees=20),   # rotation aléatoire ±20°
    transforms.ColorJitter(brightness=0.2),  # variation de luminosité ±20%
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
])

print("Transforms définis.")
```

## 5. Dataset PyTorch

Contrairement au notebook MLP, les images sont chargées **depuis le disque à chaque accès** — pas de features pré-calculées.

```python
class ImageDataset(Dataset):
    """Dataset PyTorch : chargement d'images depuis le disque à chaque accès."""

    def __init__(self, paths, labels, transform):
        # Conversion en str pour compatibilité avec Image.open (Path non accepté par toutes les versions PIL)
        self.paths     = [str(p) for p in paths]
        self.labels    = torch.LongTensor(labels)  # LongTensor requis par nn.CrossEntropyLoss
        self.transform = transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, idx):
        # Lecture de l'image depuis le disque et conversion RGB
        # (convert("RGB") garantit 3 canaux même si l'image est en niveaux de gris)
        image = Image.open(self.paths[idx]).convert("RGB")
        # Application du pipeline de transformation (resize, crop, normalisation...)
        image = self.transform(image)
        return image, self.labels[idx]


# --- Construction des 3 datasets d'entraînement ---

# Dataset A : 80 images fortement labelisées (vrais labels humains)
dataset_strong = ImageDataset(meta_strong["path"].values, labels_strong, transform_standard)

# Dataset B : 1486 images avec mix de vrais labels et weak labels DBSCAN
dataset_all = ImageDataset(meta_train["path"].values, labels_all_train, transform_standard)

# Dataset C — partie augmentée : 80 images × N_AUG = 720 copies avec transform aléatoire
# np.tile répète le tableau N_AUG fois → mêmes images renvoyées N_AUG fois avec des augmentations différentes
dataset_strong_aug = ImageDataset(
    np.tile(meta_strong["path"].values, N_AUG),   # 80 × 9 = 720 chemins
    np.tile(labels_strong, N_AUG),                # mêmes vrais labels répétés
    transform_aug                                  # transform aléatoire → image différente à chaque accès
)
# Fusion des 1486 images DBSCAN + 720 images augmentées → 2206 images total
dataset_all_aug = ConcatDataset([dataset_all, dataset_strong_aug])

# Dataset test : aucune augmentation — transform standard uniquement
dataset_test = ImageDataset(meta_test["path"].values, labels_test, transform_standard)

# --- DataLoaders : itérateurs batch pour l'entraînement ---
# shuffle=True en train pour éviter que le modèle apprenne l'ordre des images
# shuffle=False en test pour garder la correspondance labels / prédictions
loader_strong  = DataLoader(dataset_strong,  batch_size=BATCH_SMALL, shuffle=True,  num_workers=0)
loader_all     = DataLoader(dataset_all,     batch_size=BATCH_LARGE, shuffle=True,  num_workers=0)
loader_all_aug = DataLoader(dataset_all_aug, batch_size=BATCH_LARGE, shuffle=True,  num_workers=0)
loader_test    = DataLoader(dataset_test,    batch_size=BATCH_SMALL, shuffle=False, num_workers=0)

print(f"Dataset A — supervisé pur      : {len(dataset_strong)} images")
print(f"Dataset B — DBSCAN weak labels : {len(dataset_all)} images")
print(f"Dataset C — DBSCAN + augmenté  : {len(dataset_all_aug)} images")
print(f"Dataset test                   : {len(dataset_test)} images")
```

## 6. Architecture CNN avec fine-tuning des 2 dernières couches

ResNet50 se compose de : `conv1` → `layer1` → `layer2` → `layer3` → **`layer4`** → `avgpool` → **`fc`**

**Stratégie adoptée :**
- `conv1`, `layer1`, `layer2`, `layer3` → **gelés** (features générales ImageNet conservées)
- `layer4` → **dégelé**, entraîné avec `LR_LAYER4` faible (adaptation au domaine médical)
- `fc` → **remplacée** par une tête MLP (2048→512→128→2), entraînée avec `LR_HEAD` plus élevé

```python
def build_cnn_classifier(hidden_dim=512, n_classes=2, dropout=0.4):
    """ResNet50 avec layer4 dégelé et tête MLP personnalisée."""

    # Chargement de ResNet50 avec poids pré-entraînés sur ImageNet (1.2M images, 1000 classes)
    model = models.resnet50(weights=models.ResNet50_Weights.IMAGENET1K_V1)

    # Gel de tous les paramètres du backbone : conv1, layer1, layer2, layer3, layer4
    # requires_grad=False → aucun gradient calculé → poids non modifiés pendant backward()
    for param in model.parameters():
        param.requires_grad = False

    # Dégel de layer4 uniquement (dernier bloc résiduel convolutif de ResNet50)
    # Adaptation des représentations de haut niveau au domaine IRM cérébral
    # LR faible appliqué dans make_optimizer pour ne pas détruire les poids ImageNet
    for param in model.layer4.parameters():
        param.requires_grad = True

    # Remplacement de la tête de classification ImageNet (2048→1000)
    # par une tête MLP adaptée à notre problème binaire (2048→hidden_dim→128→2)
    # Toute nouvelle couche nn.* a requires_grad=True par défaut
    model.fc = nn.Sequential(
        nn.Linear(2048, hidden_dim),   # projection depuis les 2048 features de avgpool
        nn.BatchNorm1d(hidden_dim),    # normalisation du batch → stabilisation de l'entraînement
        nn.ReLU(),                     # activation non-linéaire
        nn.Dropout(dropout),           # régularisation → réduit le sur-apprentissage

        nn.Linear(hidden_dim, 128),    # compression vers 128 neurones
        nn.BatchNorm1d(128),
        nn.ReLU(),
        nn.Dropout(dropout * 0.75),    # dropout légèrement réduit dans la couche plus petite

        nn.Linear(128, n_classes)      # sortie : 2 logits (cancer / normal), sans softmax ici
        # softmax appliqué dans CrossEntropyLoss (plus stable numériquement)
    )

    return model


def make_optimizer(model, lr_head, lr_layer4):
    """Optimizer Adam avec groupes de paramètres pour LR différencié."""
    # Deux groupes distincts → deux vitesses d'apprentissage
    # layer4 : LR faible car poids pré-entraînés à préserver
    # fc     : LR plus élevé car initialisation aléatoire, apprentissage depuis zéro
    return torch.optim.Adam([
        {"params": model.layer4.parameters(), "lr": lr_layer4},
        {"params": model.fc.parameters(),     "lr": lr_head},
    ])


# --- Vérification des paramètres entraînables ---
model_test  = build_cnn_classifier(HIDDEN_DIM, N_CLASSES, DROPOUT)
n_total     = sum(p.numel() for p in model_test.parameters())
n_trainable = sum(p.numel() for p in model_test.parameters() if p.requires_grad)
n_layer4    = sum(p.numel() for p in model_test.layer4.parameters())
n_fc        = sum(p.numel() for p in model_test.fc.parameters())

print(f"Paramètres totaux       : {n_total:,}")
print(f"Paramètres entraînables : {n_trainable:,}")
print(f"  dont layer4           : {n_layer4:,}")
print(f"  dont tête fc          : {n_fc:,}")
print(f"Paramètres gelés        : {n_total - n_trainable:,}  (conv1, layer1, layer2, layer3)")
print(f"\nTête fc :")
print(model_test.fc)
```

## 7. Fonctions d'entraînement et d'évaluation

```python
def train_model(model, loader, optimizer, criterion, device, n_epochs, desc=""):
    """Boucle d'entraînement standard : forward → loss → backward → mise à jour des poids."""
    model.train()  # activation du mode entraînement (Dropout et BatchNorm actifs)
    losses = []

    for epoch in range(n_epochs):
        epoch_loss = 0.0

        for images, labels in loader:
            # Transfert des tenseurs vers GPU (ou CPU si DEVICE="cpu")
            images = images.to(device)
            labels = labels.to(device)

            optimizer.zero_grad()               # remise à zéro des gradients accumulés du batch précédent
            outputs = model(images)             # forward pass : image → ResNet50 → tête fc → logits
            loss    = criterion(outputs, labels)# CrossEntropyLoss : compare logits aux vrais labels
            loss.backward()                     # rétropropagation : calcul du gradient de la loss
            # pour chaque paramètre trainable (layer4 + fc)
            optimizer.step()                    # descente de gradient : mise à jour des poids

            epoch_loss += loss.item()           # .item() : extraction de la valeur scalaire du tenseur

        avg_loss = epoch_loss / len(loader)     # moyenne de la loss sur tous les batchs
        losses.append(avg_loss)

        if (epoch + 1) % 10 == 0:
            print(f"  {desc} Epoch {epoch+1:3d}/{n_epochs} — Loss : {avg_loss:.4f}")

    return losses  # historique des losses pour tracer la courbe d'apprentissage


def evaluate_model(model, loader, device):
    """Inférence sur un dataset : retourne vrais labels, prédictions et probabilités."""
    model.eval()  # désactivation de Dropout et passage en mode inférence de BatchNorm
    all_preds, all_labels, all_probas = [], [], []

    with torch.no_grad():  # désactivation du calcul de gradient (mémoire GPU réduite, inférence plus rapide)
        for images, labels in loader:
            images  = images.to(device)
            outputs = model(images)                                     # logits bruts (avant softmax)
            probas  = torch.softmax(outputs, dim=1)[:, 1].cpu().numpy()# probabilité de la classe cancer (index 1)
            preds   = torch.argmax(outputs, dim=1).cpu().numpy()        # classe prédite = argmax des logits
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())
            all_probas.extend(probas)

    return np.array(all_labels), np.array(all_preds), np.array(all_probas)


def print_metrics(y_true, y_pred, y_proba, model_name):
    """Calcul et affichage des 3 métriques principales + rapport de classification."""
    acc = accuracy_score(y_true, y_pred)
    # F2-score macro : β=2 donne un poids double au rappel vs la précision
    # → pénalise davantage les faux négatifs (cancer non détecté)
    f2  = fbeta_score(y_true, y_pred, beta=2, average="macro")
    # ROC-AUC : capacité de séparation des classes indépendamment du seuil de décision
    auc = roc_auc_score(y_true, y_proba)

    print(f"\n{'='*40}")
    print(f"Modèle : {model_name}")
    print(f"  Accuracy  : {acc:.4f}")
    print(f"  F2-score  : {f2:.4f}")
    print(f"  ROC-AUC   : {auc:.4f}")
    print(f"{'='*40}")
    # Rapport détaillé par classe : précision, rappel, F1
    print(classification_report(y_true, y_pred, target_names=["normal", "cancer"]))
    return acc, f2, auc


print("Fonctions définies.")
```

## 8. Recherche des hyperparamètres avec Optuna

**Stratégie :** une seule étude Optuna couvrant **tous les HP en même temps** (lr_head, lr_layer4, hidden_dim, dropout, epochs).

- **Algorithme** : TPE (Tree-structured Parzen Estimator) — bayésien, apprend des essais précédents
- **Validation** : StratifiedKFold(k=3) sur les 80 images fortement labelisées, entraînement CNN complet sur images (GPU)
- **Pruning** : MedianPruner — interrompt un essai après chaque fold si son F2 est sous la médiane des essais terminés
- **Métrique** : F2-score macro (rappel pondéré ×2, contexte médical)
- **Espace de recherche** : 5 hyperparamètres, continu + discret

### Pré-chargement des 80 images en mémoire GPU

Optimisation critique : chargement des 80 images une seule fois en mémoire GPU avant la boucle Optuna.  
Sans cela, chaque trial recharge toutes les images depuis le disque à chaque epoch (goulot d'étranglement I/O).  
Gain attendu : ×5 à ×10 sur la durée totale du HP tuning.

```python
from torch.utils.data import TensorDataset

# Chargement et transformation des 80 images une seule fois (CPU)
print("Pré-chargement des 80 images...")
tensors_strong = torch.stack([
    transform_standard(Image.open(p).convert("RGB"))
    for p in meta_strong["path"].values
])  # shape : (80, 3, 224, 224)

labels_strong_t = torch.LongTensor(labels_strong)

# Transfert vers GPU une seule fois — toutes les images résident en mémoire GPU
# (~48 Mo, négligeable pour un GPU moderne)
tensors_strong  = tensors_strong.to(DEVICE)
labels_strong_t = labels_strong_t.to(DEVICE)

print(f"Images pré-chargées sur {DEVICE} : {tensors_strong.shape}")
print(f"Mémoire GPU utilisée : ~{tensors_strong.element_size() * tensors_strong.nelement() / 1e6:.1f} Mo")
```

### 8a. Définition de l'objectif Optuna

```python
import optuna
from optuna.samplers import TPESampler
from sklearn.model_selection import StratifiedKFold

# Suppression des logs verbeux d'Optuna (un affichage par trial suffit)
optuna.logging.set_verbosity(optuna.logging.WARNING)

# Indices des 80 images fortement labelisées — base de la validation croisée
idx_strong_all = np.arange(len(meta_strong))

# Division en 3 folds stratifiés : chaque fold garde les proportions cancer/normal
skf_optuna = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)


def objective(trial):
    """Fonction objectif appelée par Optuna à chaque trial.

    Retourne le F2-score moyen sur 3 folds de validation croisée.
    Optuna cherche à maximiser cette valeur.
    """
    # --- Espace de recherche : Optuna propose une valeur pour chaque HP ---
    # suggest_float(..., log=True) : échelle logarithmique adaptée aux learning rates
    lr_head    = trial.suggest_float("lr_head",    1e-4, 5e-3, log=True)
    lr_layer4  = trial.suggest_float("lr_layer4",  1e-6, 1e-3, log=True)
    # suggest_categorical : choix discret parmi une liste
    hidden_dim = trial.suggest_categorical("hidden_dim", [256, 512, 1024])
    # suggest_float sans log=True : distribution uniforme
    dropout    = trial.suggest_float("dropout", 0.15, 0.50)
    # suggest_int : entier uniforme entre 20 et 100
    n_epochs   = trial.suggest_int("epochs", 20, 100)

    fold_f2 = []  # accumulation des scores de chaque fold

    for fold, (train_idx, val_idx) in enumerate(skf_optuna.split(idx_strong_all, labels_strong)):

        # Découpage des tenseurs GPU directement par index de fold
        # Pas de chargement disque : les images sont déjà en mémoire GPU (pré-chargées avant la boucle)
        X_tr = tensors_strong[train_idx]    # tenseur (n_train, 3, 224, 224) sur GPU
        y_tr = labels_strong_t[train_idx]   # tenseur (n_train,) sur GPU
        X_vl = tensors_strong[val_idx]
        y_vl = labels_strong_t[val_idx]

        # TensorDataset + DataLoader : itération par batchs sur des tenseurs GPU
        dl_tr = DataLoader(TensorDataset(X_tr, y_tr), batch_size=16, shuffle=True)

        # Instanciation d'un nouveau modèle pour ce fold (poids indépendants)
        m      = build_cnn_classifier(hidden_dim, N_CLASSES, dropout).to(DEVICE)
        opt    = make_optimizer(m, lr_head=lr_head, lr_layer4=lr_layer4)
        loss_fn = nn.CrossEntropyLoss()

        # Entraînement complet sur le fold train
        # imgs et lbls sont déjà sur GPU → pas de .to(DEVICE) nécessaire
        m.train()
        for _ in range(n_epochs):
            for imgs, lbls in dl_tr:
                opt.zero_grad()
                loss_fn(m(imgs), lbls).backward()  # calcul du gradient
                opt.step()                         # mise à jour des poids

        # Évaluation sur le fold val en un seul forward pass (pas de DataLoader nécessaire)
        m.eval()
        with torch.no_grad():
            preds = torch.argmax(m(X_vl), dim=1).cpu().numpy()
        fold_f2.append(fbeta_score(y_vl.cpu().numpy(), preds, beta=2, average="macro"))

        # Signal de pruning envoyé à Optuna après chaque fold
        # Optuna compare la moyenne courante à la médiane des trials précédents
        # → interromption si ce trial est clairement sous la médiane
        trial.report(float(np.mean(fold_f2)), step=fold)
        if trial.should_prune():
            raise optuna.exceptions.TrialPruned()  # arrêt immédiat du trial

    # Retour du score moyen sur les 3 folds — valeur que TPE cherche à maximiser
    return float(np.mean(fold_f2))


print("Fonction objectif définie (version optimisée — images en mémoire GPU).")
print(f"Espace de recherche :")
print(f"  lr_head    : log-uniform [1e-4, 5e-3]")
print(f"  lr_layer4  : log-uniform [1e-6, 1e-3]")
print(f"  hidden_dim : {{256, 512, 1024}}")
print(f"  dropout    : uniform [0.15, 0.50]")
print(f"  epochs     : int [20, 100]")
```

### 8b. Lancement de l'étude Optuna (50 trials, GPU)

```python
# TPESampler : algorithme bayésien (Tree-structured Parzen Estimator)
# → modélisation de la distribution des bons HP à partir des trials précédents
# seed=RANDOM_STATE pour la reproductibilité
sampler = TPESampler(seed=RANDOM_STATE)

# MedianPruner : interrompt un trial si son score intermédiaire est sous la médiane
# n_startup_trials=5 : aucun pruning avant 5 trials terminés (médiane non significative avant)
# n_warmup_steps=1   : attend au moins 1 fold avant de décider de pruner
pruner = optuna.pruners.MedianPruner(n_startup_trials=5, n_warmup_steps=1)

# Création de l'étude Optuna
# direction="maximize" : chercher le F2-score le plus élevé
study = optuna.create_study(
    direction="maximize",
    sampler=sampler,
    pruner=pruner,
    study_name="cnn_hp_tuning"
)

print(f"Appareil utilisé : {DEVICE}")
print(f"Lancement de 50 trials Optuna...\n")

# Lancement de l'optimisation : 50 appels à objective()
# show_progress_bar=True : barre de progression dans le notebook
study.optimize(objective, n_trials=50, show_progress_bar=True)

# Résumé de l'étude
print(f"\nEssais terminés : {len(study.trials)}")
print(f"Essais pruned   : {sum(1 for t in study.trials if t.state == optuna.trial.TrialState.PRUNED)}")
print(f"\nMeilleur trial  : #{study.best_trial.number}")
print(f"Meilleur F2     : {study.best_trial.value:.4f}")
print(f"\nMeilleurs HP :")
for k, v in study.best_params.items():
    print(f"  {k:12s} = {v}")
```

### 8c. Extraction des meilleurs HP et visualisations

```python
# Extraction des meilleurs HP depuis study.best_params
# → variables globales utilisées par les modèles A, B et C
BEST_LR_HEAD   = study.best_params["lr_head"]
BEST_LR_LAYER4 = study.best_params["lr_layer4"]
BEST_HIDDEN    = study.best_params["hidden_dim"]
BEST_DROPOUT   = study.best_params["dropout"]
BEST_EPOCHS_A  = study.best_params["epochs"]

print(f"HP retenus pour les modèles A / B / C :")
print(f"  lr_head    = {BEST_LR_HEAD:.2e}")
print(f"  lr_layer4  = {BEST_LR_LAYER4:.2e}")
print(f"  hidden_dim = {BEST_HIDDEN}")
print(f"  dropout    = {BEST_DROPOUT:.2f}")
print(f"  epochs     = {BEST_EPOCHS_A}")

# Tableau récapitulatif de tous les trials terminés (pruned exclus), trié par F2 décroissant
df_trials = (
    study.trials_dataframe()
    .query("state == 'COMPLETE'")
    .sort_values("value", ascending=False)
)
cols = ["number", "value", "params_lr_head", "params_lr_layer4",
        "params_hidden_dim", "params_dropout", "params_epochs"]
print(f"\nTop 10 des trials terminés :")
print(df_trials[cols].head(10).to_string(index=False))
```

### Diagnostic de sur-apprentissage

Évaluation de chaque modèle sur les **80 images d'entraînement** (phase finale pour B et C) et comparaison avec les métriques test.

- **Gap faible** (< 0.10) → bonne généralisation
- **Gap élevé** (≥ 0.10) → signal de sur-apprentissage

```python
# Diagnostic de sur-apprentissage : évaluation train vs test pour A, B et C
# Référence train = 80 images d'entraînement (phase 2 pour B et C)
# Gap = métrique_train - métrique_test → gap élevé = sur-apprentissage

rows_overfit = []

for model, name, f2_test, auc_test in [
    (model_A, "A — Supervisé pur",    f2_A, auc_A),
    (model_B, "B — Semi-sup. DBSCAN", f2_B, auc_B),
    (model_C, "C — Semi-sup. +aug",   f2_C, auc_C),
]:
    # Inférence sur les 80 images d'entraînement (même loader que la phase finale)
    y_true_tr, y_pred_tr, y_proba_tr = evaluate_model(model, loader_strong, DEVICE)

    f2_tr  = fbeta_score(y_true_tr, y_pred_tr, beta=2, average="macro")
    auc_tr = roc_auc_score(y_true_tr, y_proba_tr)

    gap_f2  = f2_tr  - f2_test
    gap_auc = auc_tr - auc_test

    # Seuil de 0.10 : au-delà, le modèle a mémorisé le train sans généraliser
    verdict = "⚠ Sur-apprentissage" if gap_f2 >= 0.10 or gap_auc >= 0.10 else "✓ Généralisation OK"

    rows_overfit.append({
        "Modèle"    : name,
        "F2 train"  : round(f2_tr,   4),
        "F2 test"   : round(f2_test, 4),
        "Gap F2"    : round(gap_f2,  4),
        "AUC train" : round(auc_tr,   4),
        "AUC test"  : round(auc_test, 4),
        "Gap AUC"   : round(gap_auc,  4),
        "Diagnostic": verdict,
    })

df_overfit = pd.DataFrame(rows_overfit)
print("Diagnostic de sur-apprentissage — train (80 images) vs test (20 images)")
print("=" * 75)
print(df_overfit.to_string(index=False))
print()
print("Rappel : test set = 20 images → AUC=1.0 possible par chance, pas forcément sur-apprentissage.")
```

## 14. Conclusions

### Différence architecturale avec la version MLP

| | Version MLP | Version CNN (ce notebook) |
|---|---|---|
| Entrée du modèle entraînable | Vecteur 2048 (features pré-extraites) | Image 224×224 |
| ResNet50 layer4 | Figé | **Dégelé — fine-tuning** |
| ResNet50 (conv1→layer3) | Figé, utilisé une seule fois | Figé, traversé à chaque epoch |
| Modèle entraînable | MLP séparé | Tête `model.fc` intégrée dans ResNet50 |
| Type de modèle | MLP sur features | CNN avec fine-tuning (Transfer learning partiel) |
| HP tuning | StratifiedKFold 5 folds sur features | Étape A (features) + Étape B (images, lr_layer4) |

### Stratégie de fine-tuning

Le fine-tuning de `layer4` (dernier bloc résiduel) est une pratique standard en transfer learning médical :
- Les couches profondes de ResNet50 encodent des features sémantiques de haut niveau (formes, textures complexes) que l'on souhaite adapter au domaine IRM cérébral.
- Un LR différencié (`lr_layer4 << lr_head`) préserve les représentations pré-entraînées tout en permettant leur adaptation.

### Bilan

- La version CNN avec fine-tuning est **architecturalement plus complète** que la version backbone gelé : elle adapte non seulement la tête de classification mais aussi les couches convolutives profondes au domaine médical.
- La version MLP reste utile comme **baseline rapide** (features pré-extraites, entraînement sur vecteurs).
- Le HP tuning en 2 étapes (tête fc sur features + lr_layer4 sur images) limite le coût computationnel tout en optimisant les hyperparamètres critiques.
